```python
import pandas as pd
import numpy as np
import os

# 1. Define paths relative to the 'notebooks' folder
input_path = "../data/raw/ai4i2020.csv"
output_path = "../data/raw/ai4i2020_messy.csv"

# 2. Check if the original file exists before proceeding
if not os.path.exists(input_path):
    print(f"Error: Could not find original dataset at {input_path}")
    print("Please ensure you run this from within the 'notebooks' directory.")
else:
    # 3. Load the original dataset
    df_original = pd.read_csv(input_path)
    df_messy = df_original.copy()
    np.random.seed(42) # Ensure reproducibility

    print(f"Loaded original data. Shape: {df_original.shape}")

    # --- INJECT MESSINESS ---

    # A. Inject Missing Values (~3%)
    cols_for_nulls = ['Type', 'Air temperature [K]', 'Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']
    for col in cols_for_nulls:
        null_indices = np.random.choice(df_messy.index, size=int(0.03 * len(df_messy)), replace=False)
        df_messy.loc[null_indices, col] = np.nan

    # B. Inject 150 Duplicate Rows
    num_duplicates = 150
    duplicate_indices = np.random.choice(df_original.index, size=num_duplicates, replace=True)
    duplicate_rows = df_original.iloc[duplicate_indices].copy()
    df_messy = pd.concat([df_messy, duplicate_rows], ignore_index=True)

    # C. Inject Physical Outliers (Extreme Values)
    # Air Temp Outliers
    air_outlier_idx = np.random.choice(df_messy.index, size=30, replace=False)
    df_messy.loc[air_outlier_idx[:15], 'Air temperature [K]'] = 999.9
    df_messy.loc[air_outlier_idx[15:], 'Air temperature [K]'] = -50.0

    # Torque Outliers
    torque_outlier_idx = np.random.choice(df_messy.index, size=30, replace=False)
    df_messy.loc[torque_outlier_idx[:15], 'Torque [Nm]'] = 500.0
    df_messy.loc[torque_outlier_idx[15:], 'Torque [Nm]'] = -150.0

    # RPM Outliers
    rpm_outlier_idx = np.random.choice(df_messy.index, size=30, replace=False)
    df_messy.loc[rpm_outlier_idx[:15], 'Rotational speed [rpm]'] = 15000
    df_messy.loc[rpm_outlier_idx[15:], 'Rotational speed [rpm]'] = -500

    # Tool Wear Outliers
    wear_outlier_idx = np.random.choice(df_messy.index, size=30, replace=False)
    df_messy.loc[wear_outlier_idx, 'Tool wear [min]'] = 9999

    # 4. Save the new messy dataset
    df_messy.to_csv(output_path, index=False)
    print(f"\nSuccess! Messy dataset saved to: {output_path}")
    print(f"New Shape: {df_messy.shape}")
```
